### I(a)
This function returns the percentage of orders delivered later than the promised time. If the zone or time block is 'all', it loops over every zone or time block. Otherwise it uses a one-item list.

```python
def late_percentage(zone, time_block, promise):
  total = 0
  late = 0
  if zone == "all":
    zone = ZONES
  else:
    zone = [zone]
  if time_block == "all":
    time_block = TIME_BLOCKS
  else:
    time_block = [time_block]
  for z in zone:
    for tb in time_block:
      orders = delivery_times(z, tb, promise, seed = 1)
      for o in orders:
        total += 1
        if o > promise:
         late += 1
  percentage = late/total*100
  return round(percentage,2)

print(late_percentage("Far West", "Fri/Sat eve", 45))
print(late_percentage("all", "Lunch", 45))
print(late_percentage("North", "all", 45))
print(late_percentage("all", "all", 45))
```

### I(b)
This code calculates the late rate for all 12 zone and time block pairs, stores each result in a list, and sorts the list from highest to lowest late rate.

```python
late_orders = []
for zone in ZONES:
  for time_block in TIME_BLOCKS:
    total = 0
    late = 0
    promise = 45
    orders = delivery_times(zone, time_block, promise, seed = 1)
    for order in orders:
      total += 1
      if order > promise:
       late += 1
    percentage = round(late/total*100, 2)
    late_orders.append((zone, time_block, percentage))
late_orders_sorted = sorted(late_orders, key=lambda item: item[2], reverse=True)
for z, tb, lp in late_orders_sorted:
    print(f"Zone: {z}\tTime Block: {tb}\tLate Orders: {lp}%")
```

###I(c)
This function returns the average delivery time. It adds up the delivery time of every order in the selected zone and time block, then divides by the number of orders. If the zone or time block is 'all', it includes all of them.

```python
def avg_time(zone, time_block, promise):
  total = 0
  time = 0
  if zone == "all":
    zone = ZONES
  else:
    zone = [zone]
  if time_block == "all":
    time_block = TIME_BLOCKS
  else:
    time_block = [time_block]
  for z in zone:
    for tb in time_block:
      orders = delivery_times(z, tb, promise, seed = 1)
      for o in orders:
        total += 1
        time += o
  avg = time/total
  return round(avg, 2)

print(avg_time("Far West", "Fri/Sat eve", 45))
print(avg_time("all", "Lunch", 45))
print(avg_time("North", "all", 45))
print(avg_time("all", "all", 45))
```

### II(a)
This function calculates the total cost of one late order. It adds the refund to the profit lost from future orders, which is the number of lost orders (churn) multiplied by the profit margin per order: cost = refund + churn × margin.

```python
def cost_late(cost):
  refund = cost['refund']
  churn = cost['churn_orders']
  margin = cost['margin']
  total_cost = refund+margin*churn
  return total_cost
print(cost_late(COSTS))
```

### II(b)
This function finds the best promise for one zone and time block. For each promise in the list, it gets the delivery times, adds the profit margin for every order, and subtracts the late-order cost for every order delivered later than that promise. It keeps the promise with the highest net profit and returns that promise and its net profit.

```python
def best_promise(zone, time_block, promises, costs):
  best_p = 0
  highest_profit = 0
  for promise in promises:
    total = 0
    late = 0
    orders = delivery_times(zone, time_block, promise, seed = 1)
    for order in orders:
      total += costs['margin']
      if order > promise:
        late += cost_late(costs)
    net_profit = round(total-late, 2)
    if highest_profit < net_profit:
      best_p = promise
      highest_profit = net_profit
  return best_p, highest_profit

promise = list(range(5, 80, 5))
best_p, highest_profit = best_promise("Far West", "Fri/Sat eve", promise, COSTS)
print(f"Best Promise: {best_p}min \tNet Profit: ${highest_profit}")
```

### Choosing the range

A promise that is too short makes most orders late, and the late costs wipe out profit. A promise that is too long means fewer customers order, so profit drops. Net profit therefore rises and then falls, so the range must be wide enough to include the peak.

I tested 5 to 75 minutes in steps of 5, which covers promises clearly too short and well beyond today's 45 minutes. If the best promise landed at either edge, I would extend the range and test again.
